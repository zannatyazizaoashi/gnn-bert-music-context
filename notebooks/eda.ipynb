# FMA preprocessing EDA
Run after `src.preprocess`. All figures use only the generated manifest and labels.

```python
from pathlib import Path
import os, json, pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
ROOT = Path.cwd(); ROOT = ROOT.parent if ROOT.name == 'notebooks' else ROOT; os.chdir(ROOT)
manifest = pd.read_csv('data/processed/manifest.csv')
label_map = json.load(open('data/processed/label_map.json'))
names = [n for n, _ in sorted(label_map.items(), key=lambda x: x[1])]
print(manifest.shape); display(manifest.head())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=manifest, x='split', ax=axes[0])
labels = np.stack(manifest.labels.map(json.loads))
sns.barplot(x=labels.sum(0), y=names, ax=axes[1])
axes[0].set_title('Processed examples by official split')
axes[1].set_title('Label frequency')
plt.tight_layout(); plt.show()
```
